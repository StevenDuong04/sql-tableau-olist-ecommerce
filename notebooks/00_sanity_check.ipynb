```python
import pandas as pd
from pathlib import Path

folder = Path("../data/raw")
for f in sorted(folder.glob("*.csv")):
    df = pd.read_csv(f)
    print(f"\n=== {f.name} ===")
    print("Shape:", df.shape)
    print("Columns:", list(df.columns))
    print("Nulls:\n", df.isnull().sum()[df.isnull().sum() > 0])
```
